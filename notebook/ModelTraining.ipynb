## 1.5 Train/Test Split

```python
x_train, x_test, y_train, y_test = train_test_split(X,Y,test_size=0.3,random_state=42,shuffle=True)
```

## 2.1 Pre-Processing

```python
categorical_features = ['city','status', 'payment_method',]
numerical_features = ['pickup_lat', 'pickup_lng', 'drop_lat', 'drop_lng', 'distance_km', 'fare_amount','pickup_time', 'drop_time', 'dif_drop_pickup', 'pickup_hour', 'pickup_month', 'pickup_day', 'weekday', 'weekend', 'rush_hour', 'distance_hour', 'squared_distance', 'delta_lat', 'delta_long', 'pickup_cluster', 'drop_cluster', 'haversine_distance']
```

```python
numeric_pipeline = Pipeline([
    ("imputer",SimpleImputer(strategy="median")),
    ("scaler",StandardScaler)
    ])
categorical_pipeline = Pipeline([
    ("imputer",SimpleImputer(strategy="most_frequent")),
      "onehot",OneHotEncoder(handle_unknown="ignore")
      ])

preprocessor_no_standard = ColumnTransformer([
    ("cat",categorical_pipeline,categorical_features)
    ])
preprocessor = ColumnTransformer([
    ("num",numeric_pipeline,numerical_features),
    ("cat",categorical_pipeline,categorical_features)
    ])
```

```python
x_train_processed_no_standard = preprocessor_no_standard.fit_transform(x_train)
x_train_processed = preprocessor.fit_transform(x_train)
x_test_processed_no_standard = preprocessor_no_standard.transform(x_test)
x_test_processed = preprocessor.transform(x_test)
```
